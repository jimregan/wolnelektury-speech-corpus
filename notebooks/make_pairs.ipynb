```python
from pathlib import Path
```

```python
AUDIO = Path("../audio")
TEXT = Path("../text")
```

```python
paired = []
```

```python
IGNORE_STEMS = """
slowka-zbior_000
"""
```

```python
for simple in AUDIO.glob("*_00[0-9].mp3"):
    if simple.stem in IGNORE_STEMS:
        continue
    realstem = simple.stem.split("_")[0]
    if (TEXT / f"{realstem}.txt").exists():
        paired.append(f"{simple.stem}.mp3\t{realstem}.txt")
    else:
        print(realstem)
```

```python
unmatched = []
for harder in AUDIO.glob("*_[0-9][0-9][0-9]_*.mp3"):
    parts = harder.stem.split("_")
    basestem = parts[0]
    number = parts[1][1:]
    if (TEXT / f"{basestem}.txt-{number}.txt").exists():
        paired.append(f"{harder.stem}.mp3\t{basestem}.txt-{number}.txt")
    elif (TEXT / f"{basestem}.txt-{parts[1]}.txt").exists():
        paired.append(f"{harder.stem}.mp3\t{basestem}.txt-{parts[1]}.txt")
    else:
        unmatched.append(harder.stem)
```
